````python
import numpy as np

rng = np.random.default_rng(42)
NIVEAUX = ["Low", "Medium", "High"]

# ---------- 1. DONNÉES FICTIVES (à remplacer plus tard par de vraies données) ----------
def creer_donnees(n=2000):
    x1 = rng.integers(1, 4, n)   # complexité du code (1-3)
    x2 = rng.integers(0, 2, n)   # ton encourageant (0/1)
    x3 = rng.integers(1, 4, n)   # niveau initial de l'étudiant (1-3)
    s = 1.0 * x3 + 0.6 * x2 - 0.9 * x1 + rng.normal(0, 0.3, n)
    y = np.where(s < 0, 0, np.where(s < 1.2, 1, 2))  # 0=Low, 1=Medium, 2=High
    X = np.stack([x1, x2, x3], axis=1).astype(float)
    return X, y

# ---------- 2. LE RÉSEAU : entrées -> couche cachée (ReLU) -> sortie (softmax) ----------
class Reseau:
    def __init__(self, n_in=3, n_cache=5, n_out=3):
        self.W1 = rng.normal(0, 0.5, (n_in, n_cache))
        self.b1 = np.full(n_cache, 0.45)          # biais de départ comme sur tes schémas
        self.W2 = rng.normal(0, 0.5, (n_cache, n_out))
        self.b2 = np.zeros(n_out)

    def avant(self, X):                            # passe avant
        self.X = X
        self.Z1 = X @ self.W1 + self.b1            # Z = B + somme(Wi * Xi)
        self.H = np.maximum(0, self.Z1)            # ReLU
        Z2 = self.H @ self.W2 + self.b2
        e = np.exp(Z2 - Z2.max(axis=1, keepdims=True))
        self.P = e / e.sum(axis=1, keepdims=True)  # softmax
        return self.P

    def perte(self, y):                            # entropie croisée
        return -np.mean(np.log(self.P[np.arange(len(y)), y] + 1e-9))

    def arriere(self, y, lr):                      # rétropropagation + descente de gradient
        n = len(y)
        dZ2 = self.P.copy()
        dZ2[np.arange(n), y] -= 1
        dZ2 /= n
        dW2, db2 = self.H.T @ dZ2, dZ2.sum(axis=0)
        dZ1 = (dZ2 @ self.W2.T) * (self.Z1 > 0)
        dW1, db1 = self.X.T @ dZ1, dZ1.sum(axis=0)
        self.W2 -= lr * dW2; self.b2 -= lr * db2
        self.W1 -= lr * dW1; self.b1 -= lr * db1

# ---------- 3. ENTRAÎNEMENT + VALIDATION ----------
X, y = creer_donnees()
coupure = int(0.8 * len(y))
X_train, y_train, X_val, y_val = X[:coupure], y[:coupure], X[coupure:], y[coupure:]

reseau = Reseau()
for epoque in range(1, 2001):
    reseau.avant(X_train)
    perte = reseau.perte(y_train)
    reseau.arriere(y_train, lr=0.2)
    if epoque % 400 == 0:
        print(f"Époque {epoque:4d} | perte = {perte:.4f}")

precision = np.mean(reseau.avant(X_val).argmax(axis=1) == y_val)
print(f"\nPrécision sur données jamais vues : {precision:.1%}")
print("Validation :", "OK, on peut déployer" if precision >= 0.85 else "insuffisant, on retourne à l'entraînement")

# ---------- 4. STRATÉGIE + DIALOGUE ----------
STRATEGIES = {
    0: ("réexpliquer", "Je vois que c'est difficile. On reprend ensemble, étape par étape."),
    1: ("simplifier",  "Tu avances bien. Voici une version plus simple avec un exemple."),
    2: ("encourager",  "Excellent travail ! Tu maîtrises ça, passons à la suite."),
}

# ---------- 5. VÉRIFICATION + SANCTION (boucle de répétition) ----------
ANGLES = ["version simplifiée", "autre exemple concret", "analogie du quotidien", "pas à pas détaillé"]

def tuteur(x1, x2, x3, max_tours=10):
    p = reseau.avant(np.array([[x1, x2, x3]]))[0]
    niveau = int(p.argmax())
    strategie, message = STRATEGIES[niveau]
    print(f"\nProbabilités : " + ", ".join(f"{n}={v:.2f}" for n, v in zip(NIVEAUX, p)))
    print(f"Niveau prédit : {NIVEAUX[niveau]} -> stratégie : {strategie}")
    print(f"Tuteur : {message}")

    comprehension = [0.45, 0.60, 0.80][niveau]    # étudiant SIMULÉ (à remplacer par un vrai quiz)
    for tour in range(1, max_tours + 1):
        score = float(np.clip(comprehension + rng.normal(0, 0.03), 0, 1))
        print(f"  Tour {tour} | score au quiz : {score:.0%}")
        if score >= 0.70:
            print("  Validé : module suivant.")
            return
        angle = ANGLES[(tour - 1) % len(ANGLES)]
        print(f"  SANCTION : on répète avec : {angle}.")
        comprehension += 0.10
    print("  Limite de répétitions atteinte : à revoir avec un humain.")

tuteur(3, 1, 1)   # code difficile, ton encourageant, niveau initial faible
````

````python
import torch

INDEX = {s["nom"]: s for s in SUJETS}
INDEX[SUJET_TEST["nom"]] = SUJET_TEST  # Ajout du sujet jamais vu à l'entraînement

def generer_tuteur(sujet_nom, niveau, strategie, angle="aucun", tour=1, temperature=0.3):
    fiche = INDEX[sujet_nom]
    prompt = formater_prompt(fiche, niveau, strategie, angle, tour)

    messages = [
        {"role": "system", "content": SYSTEME},
        {"role": "user", "content": prompt},
    ]

    texte = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    entree = tokenizer(texte, return_tensors="pt").to(trainer.model.device)

    trainer.model.eval()
    with torch.no_grad():
        sortie = trainer.model.generate(
            **entree,
            max_new_tokens=300,
            do_sample=True,
            temperature=temperature,
            pad_token_id=tokenizer.eos_token_id
        )

    reponse_brute = tokenizer.decode(sortie[0][entree["input_ids"].shape[1]:], skip_special_tokens=True)

    # Filet de sécurité anti-hallucination : vérifie si le modèle respecte la fiche
    fiche_texte = fiche['concept'] + fiche['analogie']
    mots_cles = fiche_texte.split()
    mots_presents = sum(1 for mot in mots_cles if mot.lower() in reponse_brute.lower())

    if mots_presents < 2:  # Si le modèle dérive trop, on bascule sur la réponse garantie par la fiche
        print("⚠️ [Filet de sécurité activé : réponse de secours de la fiche]")
        return formater_reponse(fiche, strategie, angle)

    return reponse_brute

# --- TEST SUR UN SUJET INÉDIT : LES TUPLES ---
print("--- TEST SUR LES TUPLES (Jamais vus à l'entraînement) ---")
print("\n[Stratégie : Simplifier]")
print(generer_tuteur("les tuples", "Débutant", "simplifier"))

print("\n[Stratégie : Réexpliquer avec l'analogie]")
print(generer_tuteur("les tuples", "Débutant", "reexpliquer", angle="analogie"))
````
